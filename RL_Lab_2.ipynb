```python
import random

machines = ['A', 'B', 'C']
epsilon = 0.2
plays = 3

N = {m: 0 for m in machines}
R = {m: 0 for m in machines}
Q = {m: 0.0 for m in machines}

for i in range(plays):
    r = random.random()

    if r < epsilon:
        machine = random.choice(machines)
    else:
        machine = max(machines, key=lambda m: Q[m])

    print("Play", i + 1)
    print("Selected Machine:", machine)

    reward = float(input("Enter reward: "))

    N[machine] += 1
    R[machine] += reward
    Q[machine] += (reward - Q[machine]) / N[machine]

    print("Updated Q-value:", Q[machine])
    print()

print("Final Results")
print("Machine\tN\tR\tQ")

for machine in machines:
    print(f"{machine}\t{N[machine]}\t{R[machine]}\t{Q[machine]:.2f}")
```
